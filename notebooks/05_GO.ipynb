# GO data

The data was previously downloaded via Entrez from NCBI together with TIGR and GO.

```python
import pandas as pd
import re
```

```python
found_results_go = pd.read_csv("../data/example_results_go.tsv", sep="\t")
```

```python
PAT_GO = re.compile(r"^(GO:\d{7}) - (.*)$")
found_results_go[["go_accession", "go_term"]] = found_results_go["go_acc"].str.extract(PAT_GO)
found_results_go = found_results_go.drop(columns="go_acc")
```

## Add this to the example data

```python
test_data = pd.ExcelFile("../../Example_Staphylococcus_aureus_Newman_TaxId_426430.xlsx").parse("Tabelle1")
```

```python
test_data = test_data.assign(wp_id=test_data["Majority protein IDs"].str.extract(r"(WP_[\d.]+)", expand=False))
```

```python
test_data = test_data.merge(found_results_go, how="left", on="wp_id")
```
